```python
# When to Use Shorthand If
# Shorthand if statements and ternary operators should be used when:

# The condition and actions are simple
# It improves code readability
# You want to make a quick assignment based on a condition
```

```python
# Short Hand If

age = int(input("Enter your age: "))
if age >= 18: print("Eligible to Vote")
```

```python
# Short Hand If ... Else

marks = int(input("Enter marks: "))
print(f"Pass with score: {marks}") if marks >= 40 else print(f"Fail with score: {marks}")
```

```python
# Assign a Value With If ... Else

salary1 = float(input("Enter first salary: "))
salary2 = float(input("Enter second salary: "))

highest_salary = salary1 if salary1 > salary2 else salary2

print("Highest Salary =", highest_salary)
```

```python
# operations Conditions on One Line
amount = float(input("Enter purchase amount: "))
discount = amount * 0.10 if amount >= 5000 else 0
print("Discount =", discount)
print("Final Amount =", amount - discount)
```

```python
# Combining Multiple Operators
age = 25
is_stdnt = False
dscnt_code = True

if (age < 18 or age > 65) and not is_stdnt or dscnt_code:
  print("Discount applies!")
```

```python
# Using Parentheses for Clarity
temperature = 25
is_raining = False
is_weekend = True

if (temperature > 20 and not is_raining) or is_weekend:
  print("Great day for outdoor activities!")
```

```python
# Multiple Conditions
marks = int(input("Enter marks: "))

print("O") if marks >= 90 else \
print("A+") if marks >= 80 else \
print("A") if marks >= 70 else \
print("B+") if marks >= 60 else \
print("F")
```

```python
# Reversed order using reversed() function
for i in reversed(range(5)):
    print(i)
```

```python
# Sorting the list in ascending order
numbers = [5, 2, 8, 1]
for num in sorted(numbers):
    print(num)
```

```python
#reverse using while loop
count = 5
while count > 0:
    print(count)
    count -= 1
```
