```python
#factorial of a number user input as 1000 then you will get an error of maximum recursion depth exceeded in comparison because the default maximum recursion depth is 1000. You can change the maximum recursion depth using sys.setrecursionlimit() function. But it is not recommended to change the maximum recursion depth because it may lead to a stack overflow error. Instead, you can use a loop to calculate the factorial of a number.
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

n = int(input("Enter a number: "))
print(factorial(n))
```

```python
#how to handle the problem of stack overflow
    #by using a loop instead of recursion
```

```python
#Recursion Tree
#Tail recursion = the recursion call is the last thing the function does

#factorial of a number using tail recursive function
def factorial(n,result=1):
    if n == 0 or n == 1:
        return result
    else:
        return factorial(n - 1, n * result)

n = int(input("Enter a number: "))
print(factorial(n))
```

```python
#sum of n numbers using recursive function
def sum(n):
    if n == 0:
        return 0
    else:
        return n + sum(n - 1)

n = int(input("Enter a number: "))
print(sum(n))
```

```python
#modify sum of n numbers using tail recursive function
def sum(n, result=0):
    if n == 0:
        return result
    else:
        return sum(n - 1, n + result)
n = int(input("Enter a number: "))  
print(sum(n))
```

```python
#fibonacci series using recursive function
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)
n = int(input("Enter a number: "))
print(fibonacci(n))
```

```python
#fibonacci using tail recursive function
def fibonacci(n, a=0, b=1):
    if n == 0:
        return a
    elif n == 1:
        return b
    else:
        return fibonacci(n - 1, b, a + b)
n = int(input("Enter a number: "))
print(fibonacci(n))
```

```python
#Important point about python
    #-unlike some languages such as scheme,python does not perform tail - call opttimization
    #therefore, tail recursive functions will still use up stack frames and can result in a stack overflow error if the recursion depth is too great.
    #
```

```python
#tail recursion vs Normal recursion
features | Normal Recursion | Tail Recursion
1. | In normal recursion, the recursive call is not the last operation performed by the function | In tail recursion, the recursive call is the last operation performed by the function    
2. | The stack frame for each recursive call is added to the call stack | The stack frame for each recursive call is not added to the call stack
3. | The function may need to perform additional operations before or after the recursive call | The function does not need to perform additional operations before or after the recursive call
```
