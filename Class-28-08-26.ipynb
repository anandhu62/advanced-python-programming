```python
#create/open a file
file = open("file.txt", "w")
print("file opened successfully")
file.close()
```

```python
#writing to a Text File
file = open("file.txt", "w")
file.write("Athul ")
file.write("Karthik")
file.close()
print("file written successfully")
```

```python
#writing multiple lines using writelines()
students = ["   Athul", "Karthik", "Rahul"]
file = open("file.txt", "w")
file.writelines(students)
```

```python
#append a data to a file
file = open("file.txt", "a")
file.write("Rahul")
file.close()
```

```python
#Reading a complete File
file = open("file.txt", "r")
data = file.read()
print(data)
file.close()
```

```python
#Reading a Specific Number of Characters
file = open("file.txt", "r")
data = file.read(8)
print(data)
file.close()
```

```python
#reading a specific line
file = open("file.txt", "r")
data = file.readline()
print(data)
file.close()
```

```python
#reading all lines into a list
file = open("file.txt", "r")
data = file.readlines()
print(data)
file.close()
```

```python
#Read a file using for loop
#only those element in the file
file = open("file.txt", "r")
for line in file:
    print(line)
file.close()
```
