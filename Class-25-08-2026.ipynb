```python
def factorial(n, level=0):
    print("    " * level + f"factorial({n})")

    if n == 1:
        print("    " * level + "return 1")
        return 1

    result = n * factorial(n - 1, level + 1)

    print("    " * level + f"return {result}")
    return result


n = int(input("Enter a number: "))

print("\nRecursion Tree and Return:")
result = factorial(n)

print(f"\nFactorial = {result}")
```

User-Defined Modules in Python

A user-defined module is a Python file created by the programmer that contains functions, variables, or classes. It can be reused in another Python program using the import statement.

Steps to create a user-defined module:

Create a Python file with a .py extension.
Define functions, variables, or classes inside the file.
Import the module into another Python program using import.
Access the functions or variables using the module name.called and the result is printed.

```python
how to create a  package in python?
To create a package in Python, follow these steps:
1. Create a new directory for your package. The name of the directory will be the name of your package.
2. Inside the package directory, create an __init__.py file. This file can be empty or contain initialization code for the package. 
3. Create additional Python files (modules) inside the package directory. These files will contain the code for your package.
```
