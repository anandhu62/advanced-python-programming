```python
'''
Looping / Iterative-Based Problems

For Loop – Fixed number of iterations
    Example: Printing numbers, sum of first N numbers, factorial calculation

While Loop – Looping until a condition becomes false
    Example: Menu-driven programs, reading input until sentinel value

Nested Loops – Loops inside loops
    Example: Multiplication tables, pattern printing, matrix operations

Common Pitfalls in Loops & Conditions:
1. Infinite Loops
2. Off-by-One Errors
3. Condition Misuse
4. Nested Loop Complexity
5. Wrong Placement of jumping statements
6. Logical Operator Mistakes

Best Practices
- Always ensure loop variables are updated in while loops.
- Check loop exit conditions carefully.
- Prefer for loops for count-based iteration and while for condition-based iteration.
- Avoid unnecessary nesting—use functions or break down logic.
- Test loops with boundary values.
'''
```

```python
#Multiplication tables
num = int(input("Enter a number: "))

for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

```python
#printing all even numbers between 1 and 50 using a for loop:
for num in range(1, 51):
    if num % 2 == 0:
        print(num)
```

```python
#Take numbers as input until 0 is entered and prints their sum
total = 0
num = int(input("Enter a number (0 to stop): "))

while num != 0:
    total += num
    num = int(input("Enter a number (0 to stop): "))

print("Sum of entered numbers:", total)
```

```python
#Factorial of a num
num = int(input("Enter a number: "))
fact = 1

for i in range(1, num + 1):
    fact *= i

print("Factorial of", num, "is", fact)
```

```python
#Print pattern
for i in range(1, 5):        
    for j in range(1, i+1): 
        print("*", end=" ")
    print()
```

```python
#Reverse pattern
for i in range(4, 0, -1):     
    for j in range(1, i+1):
        print("*", end=" ")
    print()
```

```python
#Nested loop to print pattern
rows = int(input("Enter rows: "))

for i in range(1, rows + 1):
    for space in range(rows - i):
        print(" ", end="")
    for star in range(i):
        print("* ", end="")
    print()
```

```python
#Matrix multiplication
rows_A = int(input("Enter rows for Matrix A: "))
cols_A = int(input("Enter columns for Matrix A: "))
rows_B = int(input("Enter rows for Matrix B: "))
cols_B = int(input("Enter columns for Matrix B: "))

if cols_A != rows_B: # Check if multiplication is possible
    print("Matrix multiplication not possible!")
else:
    print("Enter elements of Matrix A:")     # Read Matrix A
    A = []
    for i in range(rows_A):
        row = []
        for j in range(cols_A):
            row.append(int(input(f"A[{i}][{j}]: ")))
        A.append(row)

    print("Enter elements of Matrix B:")    # Read Matrix B
    B = []
    for i in range(rows_B):
        row = []
        for j in range(cols_B):
            row.append(int(input(f"B[{i}][{j}]: ")))
        B.append(row)

    # Initialize Result Matrix with zeros
    result = []
    for i in range(rows_A):
        row = []
        for j in range(cols_B):
            row.append(0)
        result.append(row)

    # Matrix multiplication logic
    for i in range(rows_A):
        for j in range(cols_B):
            for k in range(cols_A):
                result[i][j] += A[i][k] * B[k][j]

    # Display Result
    print("\nResultant Matrix:")
    for row in result:
        print(row)
```

```python
#Pascal triangle
n = int(input("Enter n to generate pascal triangle: "))# Number of rows
pascal = []
for i in range(n):
    row = []
    for j in range(n):
        row.append(0)
    pascal.append(row)
    
for i in range(n):
    pascal[i][0] = 1
    #print("outer loop i:",i)
    for j in range(1, i + 1):
        pascal[i][j] = pascal[i-1][j-1] + pascal[i-1][j]

print("Pascal's Triangle:")
for i in range(n):
    for j in range(i + 1):
        print(pascal[i][j], end=" ")
    print()
```

```python
# Multiplication table using nested loops
table = []
n = int(input("Enter number for multiplication table generation: "))
for i in range(1, n):       # rows
    row = []
    for j in range(1, n):   # columns
        row.append(i * j)
    table.append(row)

# Displaying the table
for row in table:
    print(row)
```
