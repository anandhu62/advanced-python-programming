```python
list=list(map(int, input().split()))
list.sort()
k=int(input())
count=0
n=len(list)
start=0
while start<n-1:
    if abs(list[start]-list[start+1])<=k:
        count+=1
        start+=2
    else:
        start+=1
print(count)
```

```python
arrival=list(map(int, input().split()))
departure=list(map(int, input().split()))
i=0
j=0
arrival.sort()
departure.sort()
platforms=0
n=len(arrival)
maximum=0
while i<n and j<n:
    if arrival[i]<=departure[j]:
        platforms+=1
        maximum=max(maximum,platforms)
        i+=1
    else:
        platforms-=1
        j+=1    
print(maximum)
```

```python
arr=list(map(int, input().split()))
k=int(input())
maximum=0
current_sum=0
prefix_sum={0:-1}
for i in range(len(arr)):
    current_sum+=arr[i]
    needed=current_sum-k
    if needed in prefix_sum:
        maximum=max(maximum,i-prefix_sum[needed])
    if current_sum not in prefix_sum:
        prefix_sum[current_sum]=i
print(maximum)

```

```python
arr=list(map(int, input().split()))
prefix={0:-1}
balance=0
maximum=0
for i in range(len(arr)):
    if arr[i]==1:
        balance+=1
    else:
        balance-=1
    if balance in prefix:
        maximum=max(maximum,i-prefix[balance])
    else:
        prefix[balance]=i
print(maximum)
```

```python
arr=list(map(int, input().split()))
k=int(input())
freq={}
left=0
maximum=0
for right in range(len(arr)):
    freq[arr[right]]=freq.get(arr[right],0)+1
    while len(freq)>k:
        freq[arr[left]]-=1
        if freq[arr[left]]==0:
            del freq[arr[left]]
        left+=1
    maximum=max(maximum,right-left+1)
print(maximum)
```

```python
n = int(input())

u = list(map(int, input().split()))
v = list(map(int, input().split()))

parent = list(range(n))

tree = True


def find(x):
    while parent[x] != x:
        x = parent[x]
    return x


def apply_union(x, y):
    parent[x] = y


# A tree must have n-1 edges
if len(u) != n - 1:
    tree = False
else:

    for i in range(len(u)):

        x = find(u[i])
        y = find(v[i])

        if x == y:
            tree = False
            break

        apply_union(x, y)


if tree:
    print("YES")
else:
    print("NO")
```

```python
import heapq
n=int(input())
m=int(input())
graph=[[] for _ in range(n)]

for _ in range(m):
    u,v,w=map(int,input().split())
    graph[u].append((v,w))

source=int(input())
dist=[float('inf')]*n
dist[source]=0
heap=[[0,source]]
while heap:
    current_dist,u=heapq.heappop(heap)
    if current_dist>dist[u]:
        continue
    for v,w in graph[u]:
        new_dist=current_dist+w
        if new_dist<dist[v]:
            dist[v]=new_dist
            heapq.heappush(heap,[new_dist,v])

print(dist)
```
