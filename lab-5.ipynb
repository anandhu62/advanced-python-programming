```python
#An Mca Student wants to store student details 
#Roll number,Name,Course,Marks

with open("student_records.txt","a") as file:
    n=int(input("Enter number of students:"))
    for i in range(n):
        roll=input("Enter Roll number:")
        name=input("Enter name:")
        course=input("Enter course:")
        marks=input("Enter marks:")
        file.write(f"{roll},{name},{course},{marks}\n")
print("Student Record Saved")
```

```python
#Reading the student record from file
with open("student_records.txt","r") as file:
    for line in file:
        roll,name,course,marks=line.strip().split(",")
        print("\nRoll number:",roll)
        print("Name:",name)
        print("Course:",course)
        print("Marks:",marks)
        
```

```python
#Search a student in a file
roll_search=input("Enter roll number to search:")
with open("student_records.txt","r") as file:
        for line in file:
              roll,name,course,marks=line.strip().split(",")
              if roll==roll_search:
                   print("Student Found")
                   print("Name:",name)
                   print("Course:",course)
                   print("Marks:",marks)
                   break
        else:
            print("Student not found")
```

```python
#Copy one file to Another
with open("student_records.txt","r") as source:
    content=source.read()
with open("student_backup.txt","w")as destination:
    destination.write(content)
print("Backup Created Successfully")
```
