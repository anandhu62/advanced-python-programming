```python
# for...else
m = int(input("Enter a start range: "))
n = int(input("Enter an end range: "))

for j in range(m, n + 1):
    if j < 2:
        continue
    for i in range(2, int(j ** 0.5) + 1):
        if j % i == 0:
            break
    else:
        print(j)
```

```python
#Pascal's Triangle
n = 5

for i in range(n):
    print(" " * (n - i), end="")
    value = 1
    for j in range(i + 1):
        print("j:",j)
        print("before computing value:",value, end="\n")
        value = value * (i - j) // (j + 1)
        print("after computing value:",value, end="\n")
    print()
```

```python
# while + break
while True:
    print("-----------\n1. Add\t2. Subtract\t3. Exit")
    choice = input("Enter choice: ")

    if choice == "+":
        a = int(input("Enter first number: "))
        b = int(input("Enter second number: "))
        print("Result =", a + b)
    elif choice == "-":
        a = int(input("Enter first number: "))
        b = int(input("Enter second number: "))
        print("Result =", a - b)
    elif choice == "#":
        print("Program terminated.")
        break
    else:
        print("Invalid choice")
        continue
```

```python
while True:
    print("-------------\n'+' Add, '-' Subtract, '# or 0' Exit")
    choice = input("Enter choice: ")
    
    match choice:
        case "+":
            a = int(input("Enter first number: "))
            b = int(input("Enter second number: "))
            print("Result =", a + b)
        case "-":
            a = int(input("Enter first number: "))
            b = int(input("Enter second number: "))
            print("Result =", a - b)
        case "#" | "0":
            print("Program terminated.")
            break
        case _: # default case
            print("Invalid choice")
```

```python
balance = 10000

while True:
    print("\n----- ATM -----\n1. Check Balance\t2. Deposit\t3. Withdraw\t4. Exit")
    choice = int(input("Enter choice: "))

    if choice == 4:
        print("Thank you for using the ATM.")
        break
    if choice not in [1, 2, 3]:
        print("Invalid choice")
        continue
    if choice == 1:
        print("Balance =", balance)
    elif choice == 2:
        amount = float(input("Enter deposit amount: "))
        if amount <= 0:
            print("Invalid amount")
            continue
        balance += amount
        print("Updated balance =", balance)
    elif choice == 3:
        amount = float(input("Enter withdrawal amount: "))
        if amount <= 0 or amount > balance:
            print("Invalid withdrawal")
            continue
        balance -= amount
        print("Updated balance =", balance)
```

```python
# Dictionary of Parts of Speech

parts_of_speech = {
    "Noun": ["teacher", "Mysuru", "computer"],
    "Pronoun": ["I", "you", "he", "they"],
    "Verb": ["teach", "go", "is"],
    "Adjective": ["good", "intelligent"],
    "Adverb": ["quickly", "very"],
    "Preposition": ["in", "on", "at", "with"],
    "Conjunction": ["and", "but", "because"],
    "Interjection": ["Oh!", "Wow!", "Great!"]
}

# Display words to the user
print("Select a word from the following list:")

for words in parts_of_speech.values():
    for word in words:
        print(word, end=", ")

word = input("\nEnter a word: ").lower()

# Match the selected word
match word:
    case "teacher" | "mysuru" | "computer":
        print("Part of Speech: Noun")
    case "i" | "you" | "he" | "they":
        print("Part of Speech: Pronoun")
    case "teach" | "go" | "is":
        print("Part of Speech: Verb")
    case "good" | "intelligent":
        print("Part of Speech: Adjective")
    case "quickly" | "very":
        print("Part of Speech: Adverb")
    case "in" | "on" | "at" | "with":
        print("Part of Speech: Preposition")
    case "and" | "but" | "because":
        print("Part of Speech: Conjunction")
    case "oh!" | "wow!" | "great!":
        print("Part of Speech: Interjection")
    case _:
        print("Word not found.")
```
