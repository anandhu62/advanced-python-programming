```python
balance=1000
print("Initial balance:", balance)
print("Enter your option:\n1. Deposit\n2. Withdraw\n3. Check Balance")
option=int(input("Enter your option (1, 2, or 3): "))
if option==1:
    try:
        amount=float(input("Enter the amount to deposit: "))
    except ValueError:
        print("Error: Please enter a valid number.")
    if amount < 0:
        print("Error: Deposit amount cannot be negative.")
    else:
        balance += amount
        print("Amount deposited successfully.")
elif option==2:
    try:
        amount=float(input("Enter the amount to withdraw: "))
    except ValueError:
        print("Error: Please enter a valid number.")
    if amount < 0:
        print("Error: Withdrawal amount cannot be negative.")
    elif amount > balance:
        print("Error: Insufficient balance.")
    else:
        balance -= amount
        print("Amount withdrawn successfully.")
elif option==3:
    print("Current balance:", balance)
else:
    print("Invalid option. Please enter 1, 2, or 3.")
```

 scenario:smart traffic light simulator
 imagine you are developing a smart traffic system for a busy city,the system stimulates the traffic signals at different roads,intersection and change there colours simulataneously
 for each intersection ns reprsesnts the north-south traffic direction and ew represents the east-west traffic direction
 each direction can have a light that is red,yellow or green.
 for example,if it is market street ns is green and ew is red





```python
# Smart Traffic Light Simulator

intersections = {
    "market_street": {
        "ns": "yellow",
        "ew": "red"
    },
    "main_road": {
        "ns": "red",
        "ew": "green"
    }
}


def change_lights(intersection):

    if intersection["ns"] == "green":
        intersection["ns"] = "yellow"
        intersection["ew"] = "red"

    elif intersection["ns"] == "yellow":
        intersection["ns"] = "red"
        intersection["ew"] = "green"

    elif intersection["ew"] == "green":
        intersection["ew"] = "yellow"
        intersection["ns"] = "red"

    elif intersection["ew"] == "yellow":
        intersection["ew"] = "red"
        intersection["ns"] = "green"

n=4
i=1
while i <= n:
    print(f"\nCycle {i}")
    for name, lights in intersections.items():
        print(f"\n{name}")
        print("Before:", lights)
        change_lights(lights)
        print("After :", lights)
    i += 1

```

```python
#CLASS 25/09/2026
#Suppose a student stands 50 m away and observes top of a building at an angle of 40,What is the height of the building

import math
dist=50
angle=40
radian=math.radians(angle)
height=dist*math.tan(radian)
print("Height of the building is:",height,'metrs')
```
