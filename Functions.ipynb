```python
'''
Variable-Length Arguments – *args
- *args allows a function to accept any number of positional arguments.
- This is useful when the number of subjects is not fixed.
'''
```

```python
# Sum of n elements using Variable-Length Arguments - *args
def calculate_sum(*numbers):
    total = 0
    for number in numbers:
        total += number
    return total

print(calculate_sum(10, 20))
print(calculate_sum(10, 20, 30, 40))
print(calculate_sum(5, 10, 15, 20, 25))
```

```python
# Average using *args
def average(*marks):
    return sum(marks) / len(marks)

print("Average =", average(85, 90, 78, 88, 92))
```

```python
'''
Variable-Length Keyword Arguments – **kwargs
- **kwargs allows a function to accept any number of keyword arguments.
'''
```

```python
#Student details using Variable-Length Keyword Arguments **kwargs
def student_details(**details):
    for key, value in details.items():
        print(key, ":", value)

student_details(name="Rahul", age=22, course="MCA", semester=2)
```

```python
#Student details using **kwargs
def employee(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

employee(name="Anita", department="IT", experience=5, salary=65000)
```

```python
'''
Combining *args and **kwargs
- A function can accept both types.
'''
```

```python
# student details using both *args and **kwargs
def student(*subjects, **details):
    print("Subjects:")
    for subject in subjects:
        print("-", subject)
    print("\nStudent Details:")
    for key, value in details.items():
        print(key, ":", value)

student("Python", "DBMS", "Machine Learning", name="Rahul", semester=2)
```

```python
# Difference between print() and return()
def add_print(a, b):
    print(a + b)

def add_return(a, b):
    return a + b

result1 = add_print(30, 100)
print("Result of print:", result1)

result2 = add_return(10, 20)
print("Result of return:", result2)
# The first function displays the result but returns None. 
# The second function returns the value, allowing it to be stored or used in another expression.
```

```python
# Returning Multiple Values
def result(*marks):
    total = sum(marks)
    percentage = total / len(marks)
    return total, percentage

#marks = [85, 90, 78, 88, 92]
total, percentage = result(85,90,99,78)

print("Total:", total)
print("Percentage:", percentage)
```

```python
'''
global Keyword
- The global keyword allows a function to modify a global variable.
'''
```

```python
# Check bank balance using the GLOBAL keyword
balance = 10000 

def deposit(amount):
    global balance
    balance += amount

deposit(5000)
print("Balance =", balance)
```

```python
def employee(**details):
    salary = details.get("salary", 0)
    experience = details.get("experience", 0)   # dictionary.get(key, default_value)
    bonus = 10000 if experience >= 5 else 5000
    final_salary = salary + bonus
    return final_salary

salary = employee(name="Anita", department="IT", experience=6, salary=60000)
print("Final Salary:", salary)
```

```python
'''
Scenario 1 - Electricity Bill Calculator
An electricity board wants to develop a Python-based billing system for 
different categories of consumers.

The billing rules are:
Consumer_Category	Rate_per_Unit
Domestic	          ₹5
Commercial	          ₹8
Heavy Load	          ₹12

Billing rule: The first 120 units are free for every consumer category. 
If consumption exceeds 120 units, the consumer is charged only for the units consumed 
beyond 120.
'''
```

```python
def calculate_bill(name, units, rate=5):
    FREE_UNITS = 120
    if units <= FREE_UNITS:
        bill = 0
    else:
        chargeable_units = units - FREE_UNITS
        bill = chargeable_units * rate
    return bill

# User input
name = input("Enter customer name: ")
units = int(input("Enter units consumed: "))
category = input("Enter category (Domestic/Commercial/Heavy Load): ").strip().lower()

if category == "domestic":         # Select rate based on category
    bill = calculate_bill(name=name, units=units, rate=5)
elif category == "commercial":
    bill = calculate_bill(name=name, units=units, rate=8)
elif category == "heavy load":
    bill = calculate_bill(name=name, units=units, rate=12)
else:
    print("Invalid consumer category")
    bill = None

if bill is not None:         # Display bill
    print("\n----- Electricity Bill -----")
    print("Customer Name:", name)
    print("Consumer Category:", category.title())
    print("Units Consumed:", units)
    print("Free Units:", 120)
    print("Bill Amount: ₹", bill)
```

```python
'''
Scenario 2: Student Performance Analyzer — *args
An MCA student may study a different number of subjects depending on the semester. 
The university wants a program that can calculate the student's percentage and grade 
regardless of the number of subjects.

Problem Statement
Develop a function using *args that:
- Accepts marks for any number of subjects.
- Calculates total and percentage.
- Determines the grade.
- Returns the percentage and grade.
'''
```

```python
def calculate_grade(*marks):
    percentage = sum(marks) / len(marks)
    grade = (
        "O" if percentage >= 90
        else "A+" if percentage >= 80
        else "A" if percentage >= 70
        else "B+" if percentage >= 60
        else "B" if percentage >= 50
        else "F"
    )
    return percentage, grade

n = int(input("Enter number of subjects: "))
marks = []

for i in range(n):
    mark = float(input(f"Enter marks for Subject {i + 1}: "))
    marks.append(mark)

percentage, grade = calculate_grade(*marks) # Function call using *list

print("\n----- Student Result -----")
print("Marks:", marks)
print("Percentage:", round(percentage, 2))
print("Grade:", grade)
```

```python
'''
Lambda Functions
- A lambda function is a small anonymous function.
- A lambda function can take any number of arguments, but can only have one expression.
- Syntax:      lambda arguments : expression
'''
```

```python
# An online store wants to calculate the discounted price of a product.
price = float(input("Enter product price: "))
discount = float(input("Enter discount percentage: "))

final_price = lambda p, d: p - (p * d / 100)

print("Final Price: ₹", final_price(price, discount))
```

```python
'''
Lambda with Built-in Functions
Lambda functions are commonly used with built-in functions like map(), filter(), and sorted().

Using Lambda with map()
The map() function applies a function to every item in an iterable

Using Lambda with filter()
The filter() function creates a list of items for which a function returns True

Using Lambda with sorted()
The sorted() function can use a lambda as a key for custom sorting

Using Lambda with reduce()
reduce() applies a function cumulatively to the elements of an iterable.
'''
```

```python
# A student has marks stored as a list. 
#The system needs to increase every mark by 5 bonus marks, subject to a maximum of 100.

marks = list(map(int, input("Enter marks separated by spaces: ").split()))
updated_marks = list(map(lambda x: min(x + 5, 100), marks))

print("Original Marks:", marks)
print("Updated Marks:", updated_marks)
```

```python
# A university provides a scholarship to students who score at least 80 marks.
marks = list(map(int, input("Enter student marks: ").split()))

eligible = list(filter(lambda mark: mark >= 80, marks))

print("All Marks:", marks)
print("Scholarship Eligible:", eligible)
```

```python
#A university has student records containing names and marks. 
#The records need to be arranged according to marks in descending order.

students = [("Suresh", 85), ("Mahesh", 72), ("Princy", 91), ("Roshin", 78)]
students = sorted(students,key=lambda student: student[1], reverse=True)
print("Students sorted by marks:")

for student in students:
    print(student)
```

```python
#An online shopping system receives the prices of multiple products and needs to calculate the total bill using reduce().

from functools import reduce

prices = list(map(float, input("Enter product prices: ").split()))

total = reduce(lambda x, y: x + y, prices)

print("Total Bill: ₹", total)
```

```python
'''
An online store has a list of product prices.
The system must:
- Add 18% GST to every product.
- Consider only products costing more than ₹500 after GST.
- Calculate the total using reduce().
'''
from functools import reduce
prices = list(map(float, input("Enter product prices: ").split()))
# Add 18% GST
prices_with_gst = list(map(lambda x: x * 1.18, prices))
# Select products above ₹500
filtered_prices = list(filter(lambda x: x > 500, prices_with_gst))
# Calculate total
total = reduce(lambda x, y: x + y,filtered_prices,0)
print("Original Prices:", prices)
print("Prices with GST:", prices_with_gst)
print("Products above ₹500:", filtered_prices)
print("Total:", round(total, 2))
```

```python
'''
Recursion
- Recursion is when a function calls itself.
- This has the benefit of meaning that you can loop through data to reach a result.
- Every recursive function must have two parts:
     - A base case - A condition that stops the recursion
     - A recursive case - The function calling itself with a modified argument
     - Without a base case, the function would call itself forever, causing a stack overflow error.
```

```python
# Factorial of a number using a recursive function
def factorial(n):
  if n == 0 or n == 1:   # Base case
    return 1
  else:                  # Recursive case
    return n * factorial(n - 1)

print(factorial(5))

'''
Recursion tree

factorial(5)
    ↓
5 × factorial(4)
          ↓
        4 × factorial(3)
                  ↓
                3 × factorial(2)
                          ↓
                        2 × factorial(1)
                                  ↓
                                  1
'''
```

```python
'''
Tail recursion = the recursive call is the last thing the function does.
'''

# factorial of a number using tail-recursive function
def factorial(n, result=1):
    if n == 0:
        return result
    return factorial(n - 1, result * n)

print(factorial(5))

'''
Recursion tree

factorial(5, 1)
       ↓
factorial(4, 5)
       ↓
factorial(3, 20)
       ↓
factorial(2, 60)
       ↓
factorial(1, 120)
       ↓
factorial(0, 120)
       ↓
      120
'''
```

```python
# sum of n numbers using recursive function
def sum_numbers(n):
    if n == 0:
        return 0
    return n + sum_numbers(n - 1)

print(sum_numbers(5))

'''
Recursion tree
   sum_numbers(5)
        |
   5 + sum_numbers(4)
        |
   4 + sum_numbers(3)
        |
   3 + sum_numbers(2)
        |
   2 + sum_numbers(1)
        |
   1 + sum_numbers(0)
        |
        0
'''
```

```python
# sum of n numbers using a tail-recursive function
def sum_numbers(n, total=0):
    if n == 0:
        return total
    return sum_numbers(n - 1, total + n)

print(sum_numbers(5))

'''
  sum_numbers(5, 0)
         |
  sum_numbers(4, 5)
         |
  sum_numbers(3, 9)
         |
  sum_numbers(2, 12)
         |
  sum_numbers(1, 14)
         |
  sum_numbers(0, 15)
         |
         15
'''
```

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

n = int(input("Enter n: "))
print("Fibonacci series: ")
for i in range(0,n+1):
    print(fibonacci(i), end=" ")

'''
                 fib(4)
                /      \
           fib(3)      fib(2)
           /   \        /   \
       fib(2) fib(1) fib(1) fib(0)
       /   \
   fib(1) fib(0)
'''
```

```python
def fibonacci(n, a=0, b=1):
    if n == 0:
        return a
    return fibonacci(n - 1, b, a + b)

n = int(input("Enter the number of terms: "))
for i in range(n):
    print(fibonacci(i), end=" ")

'''
    fib(5,0,1)
         |
    fib(4,1,1)
         |
    fib(3,1,2)
         |
    fib(2,2,3)
         |
    fib(1,3,5)
         |
    fib(0,5,8)
'''
```

```python
'''
Important point about Python
- Unlike some languages such as Scheme, Python does not perform tail-call optimization.
- Therefore, tail recursion in Python does not automatically eliminate the recursion stack.

- So, in Python, an iterative solution is generally preferable for large recursive computations.
'''

def count(n):
    if n == 0:
        return
    print(n)
    count(n - 1)

count(10000)
```

```python
'''
Tail Recursion vs Normal Recursion
Feature	                            Normal Recursion	       Tail Recursion
Recursive call	                    Not necessarily last	   Last operation
Pending operation after call	    Yes	                       No
Uses accumulator	                Usually no	               Often yes
Tail-call optimization in Python	No	                       No
Stack limitation	                Yes	                       Still yes in Python
'''
```

```python
def factorial_iterative(n):
    accumulator = 1
    while n > 0:
        accumulator *= n
        n -= 1
    return accumulator

# Example usage
print(factorial_iterative(5))  # Output: 120
```

```python
def factorial(n, level=0):
    print("    " * level + f"factorial({n})")
    if n <= 1:
        print("    " * level + "return 1")
        return 1
    result = n * factorial(n - 1, level + 1)
    print("    " * level + f"return {result}")
    return result

n = int(input("Enter a number: "))

print("\nRecursion Tree and Return:")
result = factorial(n)

print("\nFactorial =", result)
```
