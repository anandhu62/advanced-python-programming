```python
def binary_search(arr, l, h, key):

    if l > h:
        return -1

    m = (l + h) // 2

    if arr[m] == key:
        return m

    elif arr[m] > key:
        return binary_search(arr, l, m - 1, key)

    else:
        return binary_search(arr, m + 1, h, key)


arr = [1, 2, 3, 4, 5]

low = 0
high = len(arr) - 1

k = 5

index = binary_search(arr, low, high, k)

print(index)
```

```python
def result(marks):
    total = sum(marks)
    percentage = total / len(marks)

    return total, percentage


marks = list(
    map(
        int,
        input("Enter the marks separated by spaces: ").split()
    )
)

total, percentage = result(marks)

print("Total marks:", total)
print("Percentage:", percentage)
```

```python
def fact(n):
    if n==1 or n==0:
        return 1
    else:
        return n*fact(n-1)
f=fact(5)
print(f)
```
